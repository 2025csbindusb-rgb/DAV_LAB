```python
# ============================================================
# DAV MINI PROJECT
# CROP FIELD ANALYSIS AND YIELD PREDICTION
# HIGH VISIBILITY VERSION
# ============================================================

# 1. IMPORT LIBRARIES

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import ipywidgets as widgets

from IPython.display import display, HTML

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import warnings
warnings.filterwarnings('ignore')

# Set visualization theme

sns.set_theme(style="whitegrid")

plt.rcParams.update({
    'figure.facecolor': 'white',
    'axes.facecolor': 'white',
    'axes.labelcolor': 'black',
    'text.color': 'black',
    'axes.titlecolor': 'black',
    'xtick.color': 'black',
    'ytick.color': 'black',
    'font.size': 11
})

print("Libraries imported successfully!")


# 2. LOAD DATASET

url = "https://raw.githubusercontent.com/Aswins10/Agricultural-Crop-Yield-in-Indian-States-Dataset/main/crop_yield.csv"

df = pd.read_csv(url)

print("\nDATASET LOADED SUCCESSFULLY")

print("Dataset Shape:", df.shape)

display(df.head())


# 3. DATA EXPLORATION

print("\nDATASET INFORMATION")

df.info()

print("\nSTATISTICAL DESCRIPTION")

display(df.describe())

print("\nCOLUMN NAMES")

print(df.columns.tolist())

print("\nMISSING VALUES")

print(df.isnull().sum())

print("\nDUPLICATE RECORDS")

print(df.duplicated().sum())


# 4. DATA PREPROCESSING

df.columns = (
    df.columns.str.strip()
    .str.replace(' ', '_')
)

df = df.drop_duplicates()

numeric_cols = [
    'Crop_Year',
    'Area',
    'Production',
    'Annual_Rainfall',
    'Fertilizer',
    'Pesticide',
    'Yield'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(
        df[col], errors='coerce'
    )

df = df.dropna(subset=['Yield'])

for col in numeric_cols:
    df[col] = df[col].fillna(
        df[col].median()
    )

for col in ['Crop', 'Season', 'State']:
    df[col] = df[col].fillna(
        df[col].mode()[0]
    )

print("\nDATA PREPROCESSING COMPLETED")

print("Remaining Missing Values:",
      df.isnull().sum().sum())

print("Final Dataset Shape:", df.shape)


# 5. STATISTICAL ANALYSIS

print("\nCROP YIELD STATISTICAL ANALYSIS")

print("Mean:", df['Yield'].mean())

print("Median:", df['Yield'].median())

print("Mode:", df['Yield'].mode()[0])

print("Variance:", df['Yield'].var())

print("Standard Deviation:", df['Yield'].std())

print("Minimum Yield:", df['Yield'].min())

print("Maximum Yield:", df['Yield'].max())

print("Total Production:", df['Production'].sum())


# 6. CROP-WISE PRODUCTION ANALYSIS

crop_production = df.groupby(
    'Crop'
)['Production'].sum().sort_values(
    ascending=False
)

print("\nTOP 10 CROPS BY PRODUCTION")

print(crop_production.head(10))

plt.figure(figsize=(10, 5))

crop_production.head(10).plot(
    kind='bar'
)

plt.title(
    "Top 10 Crops by Production",
    color='black',
    fontweight='bold'
)

plt.xlabel("Crop", color='black')
plt.ylabel("Total Production", color='black')

plt.xticks(rotation=45, color='black')
plt.yticks(color='black')

plt.tight_layout()
plt.show()


# 7. STATE-WISE PRODUCTION ANALYSIS

state_production = df.groupby(
    'State'
)['Production'].sum().sort_values(
    ascending=False
)

plt.figure(figsize=(12, 5))

state_production.head(10).plot(
    kind='bar'
)

plt.title(
    "Top 10 States by Agricultural Production",
    color='black',
    fontweight='bold'
)

plt.xlabel("State", color='black')
plt.ylabel("Total Production", color='black')

plt.xticks(rotation=45, color='black')
plt.yticks(color='black')

plt.tight_layout()
plt.show()


# 8. SEASON-WISE PRODUCTION ANALYSIS

season_production = df.groupby(
    'Season'
)['Production'].sum()

plt.figure(figsize=(8, 5))

season_production.plot(kind='bar')

plt.title(
    "Season-wise Crop Production",
    color='black',
    fontweight='bold'
)

plt.xlabel("Season", color='black')
plt.ylabel("Total Production", color='black')

plt.xticks(rotation=45, color='black')

plt.tight_layout()
plt.show()


# 9. RAINFALL VS CROP YIELD

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='Annual_Rainfall',
    y='Yield',
    alpha=0.5
)

plt.title(
    "Rainfall vs Crop Yield",
    color='black',
    fontweight='bold'
)

plt.xlabel("Annual Rainfall (mm)", color='black')
plt.ylabel("Crop Yield", color='black')

plt.tight_layout()
plt.show()


# 10. YEAR-WISE PRODUCTION ANALYSIS

year_production = df.groupby(
    'Crop_Year'
)['Production'].sum()

plt.figure(figsize=(10, 5))

plt.plot(
    year_production.index,
    year_production.values,
    marker='o'
)

plt.title(
    "Year-wise Agricultural Production",
    color='black',
    fontweight='bold'
)

plt.xlabel("Year", color='black')
plt.ylabel("Total Production", color='black')

plt.grid(True)
plt.tight_layout()
plt.show()


# 11. CORRELATION HEATMAP

plt.figure(figsize=(10, 7))

numeric_df = df.select_dtypes(
    include=np.number
)

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    annot_kws={'color': 'black'}
)

plt.title(
    "Agricultural Data Correlation Heatmap",
    color='black',
    fontweight='bold'
)

plt.tight_layout()
plt.show()


# 12. CROP YIELD DISTRIBUTION

plt.figure(figsize=(8, 5))

sns.histplot(
    df['Yield'],
    bins=30,
    kde=True
)

plt.title(
    "Distribution of Crop Yield",
    color='black',
    fontweight='bold'
)

plt.xlabel("Crop Yield", color='black')
plt.ylabel("Frequency", color='black')

plt.tight_layout()
plt.show()


# 13. PREPARE MACHINE LEARNING FEATURES

# Production is excluded to avoid target leakage

X = df[
    [
        'Crop',
        'Crop_Year',
        'Season',
        'State',
        'Area',
        'Annual_Rainfall',
        'Fertilizer',
        'Pesticide'
    ]
]

y = df['Yield']

print("\nINPUT FEATURES")

display(X.head())

print("\nTARGET VARIABLE")

display(y.head())


# 14. SPLIT DATASET

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTRAINING RECORDS:", len(X_train))
print("TESTING RECORDS:", len(X_test))


# 15. RANDOM FOREST REGRESSION MODEL

num_features = [
    'Crop_Year',
    'Area',
    'Annual_Rainfall',
    'Fertilizer',
    'Pesticide'
]

cat_features = [
    'Crop',
    'Season',
    'State'
]

preprocessor = ColumnTransformer(
    transformers=[
        (
            'categorical',
            OneHotEncoder(handle_unknown='ignore'),
            cat_features
        ),
        (
            'numerical',
            'passthrough',
            num_features
        )
    ]
)

model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    (
        'regressor',
        RandomForestRegressor(
            n_estimators=100,
            random_state=42,
            n_jobs=-1
        )
    )
])

model.fit(X_train, y_train)

print("\nRANDOM FOREST MODEL TRAINED SUCCESSFULLY!")


# 16. PREDICT CROP YIELD

y_pred = model.predict(X_test)

print("\nACTUAL YIELD")
print(y_test.head(10).values)

print("\nPREDICTED YIELD")
print(y_pred[:10])


# 17. MODEL EVALUATION

mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)

print("\nMODEL EVALUATION RESULTS")

print("MAE:", round(mae, 3))
print("MSE:", round(mse, 3))
print("RMSE:", round(rmse, 3))
print("R2 SCORE:", round(r2, 3))


# 18. ACTUAL VS PREDICTED GRAPH

plt.figure(figsize=(8, 5))

plt.scatter(
    y_test,
    y_pred,
    alpha=0.5
)

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    'r--'
)

plt.title(
    "Actual vs Predicted Crop Yield",
    color='black',
    fontweight='bold'
)

plt.xlabel("Actual Crop Yield", color='black')
plt.ylabel("Predicted Crop Yield", color='black')

plt.tight_layout()
plt.show()


# 19. FEATURE IMPORTANCE

feature_names = model.named_steps[
    'preprocessor'
].get_feature_names_out()

importance = model.named_steps[
    'regressor'
].feature_importances_

importance_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importance
})

importance_df = importance_df.sort_values(
    by='Importance',
    ascending=False
)

print("\nTOP 10 IMPORTANT FEATURES")

display(importance_df.head(10))

plt.figure(figsize=(10, 6))

sns.barplot(
    data=importance_df.head(10),
    x='Importance',
    y='Feature'
)

plt.title(
    "Top 10 Important Agricultural Features",
    color='black',
    fontweight='bold'
)

plt.xlabel("Importance", color='black')
plt.ylabel("Feature", color='black')

plt.tight_layout()
plt.show()


# 20. HIGH-VISIBILITY INTERACTIVE DASHBOARD

# Plotly white theme with black text

px.defaults.template = "plotly_white"

px.defaults.color_discrete_sequence = [
    '#2563EB',
    '#16A34A',
    '#EA580C',
    '#9333EA',
    '#DB2777'
]

# Dashboard heading

display(HTML("""
<div style="
background-color:#FFFFFF;
padding:20px;
border:2px solid #1F2937;
border-radius:12px;
text-align:center;
margin-bottom:15px;">

<h1 style="
color:#000000;
font-size:28px;
font-weight:900;
margin:5px;">

CROP FIELD ANALYSIS DASHBOARD

</h1>

<p style="
color:#333333;
font-size:16px;
font-weight:bold;">

Agricultural Data Analysis and Yield Prediction

</p>
</div>
"""))


# Create filters

crop_filter = widgets.Dropdown(
    options=['All'] + sorted(
        df['Crop'].unique().tolist()
    ),
    value='All',
    description='Crop:',
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='350px')
)

state_filter = widgets.Dropdown(
    options=['All'] + sorted(
        df['State'].unique().tolist()
    ),
    value='All',
    description='State:',
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='350px')
)

season_filter = widgets.Dropdown(
    options=['All'] + sorted(
        df['Season'].unique().tolist()
    ),
    value='All',
    description='Season:',
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='350px')
)


# Dashboard function

def create_dashboard(crop, state, season):

    data = df.copy()

    if crop != 'All':
        data = data[data['Crop'] == crop]

    if state != 'All':
        data = data[data['State'] == state]

    if season != 'All':
        data = data[data['Season'] == season]

    if data.empty:
        display(HTML("""
        <h2 style="color:#000000;">
        No matching records found.
        </h2>
        """))
        return

    # Calculate dashboard values

    total_production = data['Production'].sum()

    average_yield = data['Yield'].mean()

    total_crops = data['Crop'].nunique()

    total_states = data['State'].nunique()


    # HIGH CONTRAST KPI CARDS

    display(HTML(f"""

    <div style="
    display:grid;
    grid-template-columns:repeat(auto-fit,minmax(170px,1fr));
    gap:15px;
    margin-top:20px;
    margin-bottom:25px;">

    <div style="
    background-color:#DBEAFE;
    border:2px solid #1D4ED8;
    padding:20px;
    border-radius:12px;
    text-align:center;">

    <h3 style="
    color:#000000;
    font-size:17px;
    font-weight:900;">

    TOTAL PRODUCTION

    </h3>

    <h1 style="
    color:#000000;
    font-size:27px;
    font-weight:900;">

    {total_production:,.0f}

    </h1>
    </div>


    <div style="
    background-color:#DCFCE7;
    border:2px solid #15803D;
    padding:20px;
    border-radius:12px;
    text-align:center;">

    <h3 style="
    color:#000000;
    font-size:17px;
    font-weight:900;">

    AVERAGE YIELD

    </h3>

    <h1 style="
    color:#000000;
    font-size:27px;
    font-weight:900;">

    {average_yield:.2f}

    </h1>
    </div>


    <div style="
    background-color:#FEF3C7;
    border:2px solid #B45309;
    padding:20px;
    border-radius:12px;
    text-align:center;">

    <h3 style="
    color:#000000;
    font-size:17px;
    font-weight:900;">

    TOTAL CROPS

    </h3>

    <h1 style="
    color:#000000;
    font-size:27px;
    font-weight:900;">

    {total_crops}

    </h1>
    </div>


    <div style="
    background-color:#F3E8FF;
    border:2px solid #7E22CE;
    padding:20px;
    border-radius:12px;
    text-align:center;">

    <h3 style="
    color:#000000;
    font-size:17px;
    font-weight:900;">

    TOTAL STATES

    </h3>

    <h1 style="
    color:#000000;
    font-size:27px;
    font-weight:900;">

    {total_states}

    </h1>
    </div>

    </div>

    """))


    # COMMON PLOTLY LAYOUT

    common_layout = dict(
        font=dict(
            family='Arial',
            size=14,
            color='#000000'
        ),
        paper_bgcolor='#FFFFFF',
        plot_bgcolor='#FFFFFF',
        title_font=dict(
            size=20,
            color='#000000'
        ),
        xaxis=dict(
            title_font=dict(color='#000000'),
            tickfont=dict(color='#000000')
        ),
        yaxis=dict(
            title_font=dict(color='#000000'),
            tickfont=dict(color='#000000')
        ),
        margin=dict(l=60, r=30, t=70, b=60)
    )


    # CHART 1: CROP-WISE PRODUCTION

    crop_data = data.groupby(
        'Crop', as_index=False
    )['Production'].sum().nlargest(
        10, 'Production'
    )

    fig1 = px.bar(
        crop_data,
        x='Crop',
        y='Production',
        title='TOP 10 CROP PRODUCTION',
        text_auto='.2s'
    )

    fig1.update_layout(**common_layout)

    fig1.show()


    # CHART 2: STATE-WISE PRODUCTION

    state_data = data.groupby(
        'State', as_index=False
    )['Production'].sum().nlargest(
        10, 'Production'
    )

    fig2 = px.bar(
        state_data,
        x='State',
        y='Production',
        title='TOP 10 STATES BY PRODUCTION'
    )

    fig2.update_layout(**common_layout)

    fig2.show()


    # CHART 3: SEASON-WISE PRODUCTION

    season_data = data.groupby(
        'Season', as_index=False
    )['Production'].sum()

    fig3 = px.pie(
        season_data,
        names='Season',
        values='Production',
        title='SEASON-WISE CROP PRODUCTION',
        hole=0.3
    )

    fig3.update_layout(
        font=dict(color='#000000', size=14),
        paper_bgcolor='#FFFFFF',
        title_font=dict(color='#000000', size=20)
    )

    fig3.update_traces(
        textfont=dict(
            color='#000000',
            size=14
        )
    )

    fig3.show()


    # CHART 4: RAINFALL VS YIELD

    fig4 = px.scatter(
        data,
        x='Annual_Rainfall',
        y='Yield',
        color='Season',
        title='RAINFALL VS CROP YIELD'
    )

    fig4.update_layout(**common_layout)

    fig4.show()


    # CHART 5: YEAR-WISE PRODUCTION

    year_data = data.groupby(
        'Crop_Year', as_index=False
    )['Production'].sum()

    fig5 = px.line(
        year_data,
        x='Crop_Year',
        y='Production',
        markers=True,
        title='YEAR-WISE CROP PRODUCTION'
    )

    fig5.update_layout(**common_layout)

    fig5.show()


# CONNECT FILTERS TO DASHBOARD

dashboard = widgets.interactive_output(
    create_dashboard,
    {
        'crop': crop_filter,
        'state': state_filter,
        'season': season_filter
    }
)

# Display filters and dashboard

display(widgets.VBox([
    crop_filter,
    state_filter,
    season_filter,
    dashboard
]))


# 21. SAMPLE CROP YIELD PREDICTION

sample = X_test.iloc[[0]].copy()

predicted_yield = model.predict(sample)[0]

actual_yield = y_test.iloc[0]

print("\nSAMPLE CROP YIELD PREDICTION")

display(sample)

print("Predicted Yield:",
      round(predicted_yield, 3))

print("Actual Yield:",
      round(actual_yield, 3))


# 22. FINAL PROJECT STATUS

display(HTML("""
<div style="
background-color:#DCFCE7;
border:2px solid #15803D;
padding:20px;
border-radius:12px;
text-align:center;">

<h2 style="
color:#000000;
font-weight:900;">

CROP FIELD ANALYSIS PROJECT COMPLETED!

</h2>

<p style="
color:#000000;
font-size:16px;
font-weight:bold;">

Data Analysis | Visualization |
Machine Learning | Interactive Dashboard

</p>
</div>
"""))

print("\nPROJECT EXECUTED SUCCESSFULLY!")
```
